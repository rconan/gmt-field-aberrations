```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.tri as tri
%matplotlib inline
```

```python
data = np.load("m12_closed-loop_calibration.pkl",allow_pickle=True)
```

```python
m1 = np.asarray(data[0]).reshape(5,-1).T
m2 = np.asarray(data[1]).reshape(5,-1).T
m12 = np.hstack([m1,m2])
u,s,vt = np.linalg.svd(m12,full_matrices=False)
```

```python

m12 = np.hstack([m1/np.linalg.norm(m1),m2/np.linalg.norm(m2)])
u,s,vt = np.linalg.svd(m12,full_matrices=False)
```

```python
fig,ax = plt.subplots()
h = ax.imshow(m12,origin="upper",cmap="RdBu")
ax.set_xticks(range(0,10))
ax.set_xticklabels(["Tx","Ty","Tz","Rx","Ry","Tx","Ty","Tz","Rx","Ry"])
ax.set_yticks(range(0,7,2))
ax.set_yticklabels([f"Z{x}" for x in range(4,11,2)])
ax.grid(alpha=0.25)
fig.colorbar(h,ax=ax)
```

```python
fig,axs = plt.subplots(ncols=2,figsize=(10,5))
ax = axs[0]
h = ax.imshow(vt.T,cmap="RdBu")
ax.set_yticks(range(0,10))
ax.set_yticklabels(["Tx","Ty","Tz","Rx","Ry","Tx","Ty","Tz","Rx","Ry"])
ax.grid(alpha=0.25)
fig.colorbar(h,ax=ax)
ax = axs[1]
ax.semilogy(s,".-")
ax.grid(alpha=0.25)
```

```python
fig,ax = plt.subplots()
ax.plot(np.cumsum(s)/np.sum(s),'.-')
ax.grid()
```
